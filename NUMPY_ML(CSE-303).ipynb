```python
#mini Project(Image Analyser)
import numpy as np
import matplotlib.pyplot as plt
# 1. Image height, width, and number of channels
height, width, channels = img_array.shape
print("Height:", height)
print("Width:", width)
print("Channels:", channels)
# 2. Average brightness
overall = np.mean(img_array)
red = np.mean(img_array[:, :, 0])
green = np.mean(img_array[:, :, 1])
blue = np.mean(img_array[:, :, 2])
print("Overall brightness:", overall)       # Display overall brightness
print("Red brightness:", red)               # Display Red brightness
print("Green brightness:", green)           # Display Green brightness
print("Blue brightness:", blue)             # Display Blue brightness

# 3. Crop the center 200x200 region
start_y = (height - 200) // 2
start_x = (width - 200) // 2

crop = img_array[start_y:start_y + 200,
                 start_x:start_x + 200]     # Crop the center
plt.imshow(crop)                            # Display cropped
plt.title("Center Crop")
plt.axis("off")                             # Hide axes
plt.show()

# 4. Flip the image vertically and horizontally
vertical_flip = img_array[::-1, :]          # Flip image vertically
horizontal_flip = img_array[:, ::-1]        # Flip image horizontally
plt.imshow(vertical_flip)                   # Display vertical flip
plt.title("Vertical Flip")
plt.axis("off")
plt.show()
plt.imshow(horizontal_flip)                 # Display horizontal flip
plt.title("Horizontal Flip")
plt.axis("off")
plt.show()
# 5. Display RGB channels separately
red_channel = img_array[:, :, 0]            # Extract Red
green_channel = img_array[:, :, 1]          # Extract Green
blue_channel = img_array[:, :, 2]           # Extract Blue
plt.imshow(red_channel, cmap="gray")        # Display Red
plt.title("Red Channel")
plt.axis("off")
plt.show()
plt.imshow(green_channel, cmap="gray")      # Display Green
plt.title("Green Channel")
plt.axis("off")
plt.show()
plt.imshow(blue_channel, cmap="gray")       # Display Blue
plt.title("Blue Channel")
plt.axis("off")
plt.show()
# 6. Convert image to black and white
gray = np.mean(img_array, axis=2)
binary = np.where(gray > 128, 255, 0)       # Above 128 = white
plt.imshow(binary, cmap="gray")             # Display black-and-white image
plt.title("Thresholded Image")
plt.axis("off")
plt.show()
```

```python
#Comparing 2 Sentences
sentence_a = "APPLE IS A FRUIT"
sentence_b = "ORANGE IS A FRUIT"
tokens_a = sentence_a.lower().split()
tokens_b = sentence_b.lower().split()
combined_vocab = np.unique(np.array(tokens_a + tokens_b))
vec_a = bag_of_words_vector(tokens_a, combined_vocab)
vec_b = bag_of_words_vector(tokens_b, combined_vocab)
print("Vector A:", vec_a)
print("Vector B:", vec_b)
cosine_sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))
print("Cosine similarity:", cosine_sim)
```

```python
#Most Frequent Words
import numpy as np
import re
text = """I'm Vinuthna. I'm studying BTech. I study at SRM University.
It is located in AP."""
tokens = re.findall(r'\b\w+\b', text.lower())
print("Tokens:", tokens)
vocabulary = sorted(set(tokens))

def bag_of_words_vector(tokens, vocabulary):
    vector = np.zeros(len(vocabulary), dtype=int)

    for word in tokens:
        index = vocabulary.index(word)
        vector[index] += 1

    return vector

word_count_vector = bag_of_words_vector(tokens, vocabulary)

print("\nVocabulary:", vocabulary)
print("Word-count vector:", word_count_vector)

sorted_indices = np.argsort(word_count_vector)[::-1]
top_3_indices = sorted_indices[:3]

print("\nTop 3 most frequent words:")
for index in top_3_indices:
    print(vocabulary[index], ":", word_count_vector[index])
```

```python
#Simple Text Similarity Checker(NLP)
import numpy as np
# Three sentences
sentence1 = "I like playing badminton"
sentence2 = "I enjoy listening songs"
sentence3 = "I like watching movies"

# Tokenize the sentences
words1 = sentence1.lower().split()
words2 = sentence2.lower().split()                #AP24110011208
words3 = sentence3.lower().split()
# Create one shared vocabulary
vocabulary = sorted(set(words1 + words2 + words3))
print("Shared Vocabulary:")
print(vocabulary)
# Create Bag-of-Words vectors
vector1 = np.array([words1.count(word) for word in vocabulary])
vector2 = np.array([words2.count(word) for word in vocabulary])
vector3 = np.array([words3.count(word) for word in vocabulary])

print("\nBag-of-Words Vectors:")
print("Sentence 1:", vector1)
print("Sentence 2:", vector2)
print("Sentence 3:", vector3)
# Cosine similarity function
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
# Calculate similarity between all pairs
similarity_12 = cosine_similarity(vector1, vector2)
similarity_13 = cosine_similarity(vector1, vector3)
similarity_23 = cosine_similarity(vector2, vector3)

print("\nCosine Similarity:")
print("Sentence 1 and Sentence 2:", similarity_12)
print("Sentence 1 and Sentence 3:", similarity_13)
print("Sentence 2 and Sentence 3:", similarity_23)
# Find the most similar pair
scores = {
    "Sentence 1 and Sentence 2": similarity_12,
    "Sentence 1 and Sentence 3": similarity_13,
    "Sentence 2 and Sentence 3": similarity_23
}
most_similar = max(scores, key=scores.get)

print("\nMost Similar Pair:")
print(most_similar)
print("Similarity Score:", scores[most_similar])
```
